```python
# Step 1: Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 2: Generate sample gene expression data
# Simulating expression data for 5 genes across 10 samples
np.random.seed(42)  # for reproducibility
genes = [f'Gene_{i}' for i in range(1, 6)]
samples = [f'Sample_{j}' for j in range(1, 11)]
data = np.random.lognormal(mean=2, sigma=0.5, size=(10, 5))
```

```python
# Step 3: Create a DataFrame
df = pd.DataFrame(data, index=samples, columns=genes)
df.to_excel('gene_expression_data.xlsx',index= False)
print("Sample Gene Expression Data:")
print(df)
```

```python
# Step 4: Compute basic descriptive statistics using describe()
desc = df.describe().T  # Transpose for easier reading
print("\nBasic Descriptive Statistics:")
print(desc[['mean', '50%', 'std', 'min', 'max']].rename(columns={'50%': 'median'}))
```

```python
# Step 9: Visualization - Boxplot for all genes
plt.figure(figsize=(8, 5))
plt.boxplot(x=df["Gene_1"], tick_labels=["Gene_1"])
plt.title('Boxplot of Gene Expression Levels')
plt.ylabel('Expression Level')
plt.grid(True)
plt.show()
```

Histogram

Shows frequency of expression values.

Good for identifying distribution shape and skewness.

```python
plt.hist(df['Gene_1'], bins=6, edgecolor='black')
plt.title('Histogram of Gene_1 Expression')
plt.xlabel('Expression Level')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
sns.kdeplot(df['Gene_1'], fill=True)
plt.title('Density Plot of Gene_1 Expression')
plt.xlabel('Expression Level')
plt.grid(True)
plt.show()
```
