# **Introduction to Sampling Techniques**

---

## **What is Inferential Statistics?**

Inferential statistics is a branch of statistics that focuses on drawing conclusions about a population based on data collected from a sample. Unlike descriptive statistics, which summarize and describe the features of a dataset, inferential statistics use probability theory to make generalizations, predictions, and decisions.

### **Key Functions of Inferential Statistics:**

1. **Estimate population parameters** (e.g., mean, proportion).
2. **Test hypotheses** using sample data.
3. **Make predictions** about future observations.
4. **Determine relationships** and associations between variables.

### **Examples:**

* Estimating the average height of adult males in a country based on a sample of 500 individuals.
* Testing whether a new drug is more effective than an existing one using data from a clinical trial.
* Predicting the probability of a gene being mutated in a cancer patient based on sample sequences.

### **Importance in Bioinformatics:**

In bioinformatics, inferential statistics allows researchers to:

* Identify differentially expressed genes between conditions (e.g., healthy vs. cancerous tissue).
* Infer functional pathways from sample-based omics data.
* Generalize findings from a small group of patients to a larger population.

---

## **Definition of Sampling**

Sampling is the process of selecting a subset of data (sample) from a larger dataset (population) to estimate population parameters. It enables researchers to make inferences about the population without analyzing every individual. This technique is fundamental in fields like bioinformatics, where analyzing the entire genome or proteome might be infeasible due to size, cost, or computational limits.

---

## **Key Concepts**

### **1. Population vs. Sample**

---

## **Population**

A **population** is the entire group of individuals, items, or data that is the subject of a study. Researchers aim to draw conclusions about the population through data collection and analysis.

### **Characteristics of a Population**

1. **Comprehensive**: Includes all relevant elements of interest.

   * **Example**: All genes in the human genome.
2. **Defined Scope**: The boundaries of the population are clearly stated.

   * **Example**: Population of diabetic patients in a hospital.
3. **May Be Large or Infinite**:

   * **Finite Population**: All students enrolled in a university.
   * **Infinite Population**: All potential outcomes of a DNA sequencing experiment.

---

## **Sample**

A **sample** is a subset of the population chosen for study. The goal of sampling is to make inferences about the population without analyzing every individual.

### **Characteristics of a Sample**

1. **Representative**: Should accurately reflect the population.
2. **Smaller in Size**: Contains fewer elements than the population.

   * **Example**: Analyzing gene expression in 200 cells out of 10,000.
3. **Practical for Analysis**: Makes data collection and processing feasible.

---

## **Key Differences Between Population and Sample**

| **Aspect**            | **Population**                           | **Sample**                                |
| --------------------- | ---------------------------------------- | ----------------------------------------- |
| **Definition**        | Entire group of interest in the study    | Subset of the population                  |
| **Size**              | Large or infinite                        | Smaller and manageable                    |
| **Representation**    | Complete                                 | Represents the population                 |
| **Examples**          | All patients in a hospital               | 100 selected patients                     |
| **Values**            | Parameters (e.g., mean: μ, variance: σ²) | Statistics (e.g., mean: x̄, variance: s²) |
| **Study Feasibility** | Expensive, time-consuming                | Cost-effective and quicker                |

---

## **Examples**

1. **Population**: All students in a university.

   * **Sample**: 200 randomly selected students.
2. **Population**: All customers of an online store.

   * **Sample**: 1,000 customers surveyed about their shopping habits.
3. **Population**: All sequences in a whole genome.

   * **Sample**: 100 randomly selected gene regions for mutation analysis.

---

## **Why Use a Sample Instead of a Population?**

1. **Feasibility**: Studying a population is often impractical due to size.
2. **Cost and Time**: Sampling reduces resource requirements.
3. **Focus**: Allows researchers to focus on specific segments of interest.

---

## **Example**

Imagine a city with a population of 1 million. To estimate the average income:

* Surveying every resident would be impractical.
* A random sample of 500 people can provide a reliable estimate of the average income with much less effort.

---

### **2. Parameter vs. Statistic**

* **Parameter**: A measurable characteristic of the population (e.g., population mean μ).

  * **Example**: The true average blood pressure of all patients in a clinic.
* **Statistic**: A measurable characteristic derived from the sample (e.g., sample mean x̄).

  * **Example**: The average blood pressure of 50 randomly chosen patients.

| Concept   | Refers To  | Symbol | Example                         |
| --------- | ---------- | ------ | ------------------------------- |
| Parameter | Population | μ, σ²  | True mean cholesterol level     |
| Statistic | Sample     | x̄, s² | Mean from 100 surveyed patients |

---

## **Importance of Sampling**

### **1. Reduces Cost and Effort**

* Collecting data from the entire population can be expensive and time-consuming.
* Sampling allows for efficient use of resources and quicker analysis.

### **2. Enables Analysis When Population Is Inaccessible**

* In cases where the entire population cannot be reached or studied, sampling provides a practical solution.
* Particularly useful in bioinformatics when data volume is immense (e.g., whole genome sequencing).

---

## **Real-World Example in Bioinformatics**

To identify differentially expressed genes in cancer:

* **Population**: All 20,000 genes in the human genome.
* **Sample**: Analyze 1,000 genes identified as most variable in expression.
* **Reason**: It's computationally intensive and biologically unnecessary to test all genes if many do not vary.

---

Sampling is an essential tool for research and analysis, enabling accurate insights with limited resources. By understanding the key concepts and importance, we can appreciate its role in deriving meaningful conclusions about larger populations, especially in fields with massive datasets like bioinformatics, epidemiology, and social sciences.


# **Types of Sampling Methods**

---

## **Probability Sampling**
In **probability sampling**, every individual in the population has a known and non-zero chance of being selected for the sample. This type of sampling ensures that the sample is representative of the population, allowing for generalization of results.

---

### **1. Simple Random Sampling**

Simple Random Sampling (SRS) is a method of sampling where each individual or element in the population has an equal chance of being selected.

**Example**:
Imagine you have a class of 50 students, and you want to randomly select 5 students for a survey. You could assign each student a unique number from 1 to 50, then randomly pick 5 numbers using a random number generator or a Python function like `sample()`.



- **Description**: Every individual has an equal chance of being selected.
- **How It Works**: Each individual in the population is randomly selected, often using a random number generator or drawing lots.
- **Example**: Randomly selecting 10 students from a class of 50.

**Advantages**:
- Simple and unbiased.
- Easy to implement with smaller populations.

**Disadvantages**:
- May require a complete list of the population (sampling frame).
- May not be efficient for large populations.

---

### **2. Stratified Sampling**

Stratified Sampling involves dividing the population into subgroups or strata based on a characteristic, then taking a random sample from each stratum. This ensures that each subgroup is well-represented in the sample.

**Example**:
If you're surveying a population of 100 people and want to make sure both males and females are represented, you could divide the population into two strata (males and females) and then randomly sample from each stratum.

- **Description**: The population is divided into distinct strata or groups based on certain characteristics, and a sample is taken from each stratum.
- **How It Works**: The strata could be based on characteristics like age, income, gender, etc. A sample is then selected randomly from each of these strata.
- **Example**: Dividing a population by age groups (18-25, 26-35, etc.) in a survey.

**Advantages**:
- Ensures representation from all subgroups.
- Provides more precise and reliable estimates.

**Disadvantages**:
- Requires detailed knowledge of the population to create strata.
- More complex to administer.

---

### **3. Cluster Sampling**

 In Cluster Sampling, the population is divided into groups or clusters (often based on geographical location or any natural grouping). Then, entire clusters are randomly selected, and all members from the selected clusters are included in the sample.

**Example**:
If you're conducting a survey of schools in a city, rather than surveying individual students, you might randomly select a few schools (clusters), and then survey all students within those selected schools.


- **Description**: The population is divided into clusters, and then a random sample of clusters is selected. Entire clusters are chosen, rather than individual members.
- **How It Works**: The clusters could be naturally occurring groups like geographical regions, schools, or organizations. Once clusters are selected, all individuals within those clusters are surveyed.
- **Example**: Selecting schools instead of individual students for a survey.

**Advantages**:
- More practical when populations are large and dispersed.
- Cost-effective in certain situations.

**Disadvantages**:
- Clusters may not be representative of the population.
- Can introduce bias if clusters are not well-matched to the population.

---

### **4. Systematic Sampling**

 Systematic Sampling involves selecting every nth element from a population after randomly selecting a starting point. This method assumes that the population is ordered in some way.

**Example**:
If you have a list of 100 people and want to select every 10th person for a survey, you would choose the first person randomly, then select every 10th individual after that. For instance, you might select the 1st, 11th, 21st, and so on.


- **Description**: Every **n-th** individual is selected from a list or sequence.
- **How It Works**: After selecting a random starting point, every **n-th** individual from the population is included in the sample.
- **Example**: Surveying every 5th person in a queue.

**Advantages**:
- Easy to implement and efficient.
- Can be useful in scenarios where a complete list of the population is available.

**Disadvantages**:
- If there's a pattern or periodicity in the population, the sample could be biased.

---

## **Non-Probability Sampling**
In **non-probability sampling**, not every individual has an equal chance of being selected. These methods are often more subjective and less representative, but they can be used when probability sampling is impractical or unnecessary.

---

### **1. Convenience Sampling**

Convenience Sampling is a non-probability sampling method where you select the sample based on ease of access, rather than a random or structured selection process. It's quick but may lead to biased results.

**Example**:
If you're conducting a survey and decide to only survey the first 10 people who walk into a store, that’s convenience sampling. You’re choosing individuals who are easiest to access.


- **Description**: Individuals are selected based on ease of access or convenience rather than randomness.
- **Example**: Surveying people who are nearby or easy to contact, such as asking friends or coworkers.

**Advantages**:
- Simple and quick.
- Low cost.

**Disadvantages**:
- High potential for bias.
- Results may not be generalizable.

---

### **2. Judgmental Sampling (Purposive Sampling)**
- **Description**: The researcher selects specific individuals based on their judgment, usually because they are believed to be representative of the population.
- **Example**: A researcher chooses experts or individuals with particular experiences for an interview.

**Advantages**:
- Useful for focused studies or expert opinions.
- Can provide insights into a specific group or behavior.

**Disadvantages**:
- High risk of researcher bias.
- Not generalizable to the broader population.

---

### **3. Snowball Sampling**
- **Description**: Initial participants are selected, and they help the researcher recruit further participants, creating a "snowball" effect.
- **Example**: Surveying people within a specific community or network, who then refer others to participate.

**Advantages**:
- Useful for studying hidden populations (e.g., drug users, or marginalized groups).
- Can help access difficult-to-reach populations.

**Disadvantages**:
- Bias in the selection of participants.
- May lead to overrepresentation of certain groups.

---


The choice of sampling method depends on the research objectives, population characteristics, and available resources. Probability sampling methods are ideal for ensuring representativeness and making generalizations, while non-probability methods are often used when resources are limited or specific, targeted information is needed.

## ✅ **1. Simple Random Sampling**

```python
import pandas as pd
from sklearn.utils import resample
import seaborn as sns
import random

# Load dataset
titanic = sns.load_dataset("titanic").dropna(subset=["age", "sex", "embarked", "class"])

# Simple random sample of 100 passengers
simple_random_sample = titanic.sample(n=100, random_state=42)

print(simple_random_sample.head())
```

**Explanation**:

* Each passenger has an equal chance of being selected.
* Useful when we want an unbiased and representative subset.



---

## ✅ **2. Stratified Sampling**


```python
from sklearn.model_selection import train_test_split

# Stratified sampling based on 'sex' column
strat_sample, _ = train_test_split(titanic,
                                   test_size=0.8,
                                   stratify=titanic['sex'],
                                   random_state=42)

print(strat_sample['sex'].value_counts(normalize=True))
```

**Explanation**:

* Ensures that both genders are proportionally represented in the sample.
* Crucial for balanced analysis (e.g., comparing survival between males and females).




---

## ✅ **3. Cluster Sampling**

```python
# Let's assume 'embarked' is our cluster (i.e., ports: C, Q, S)
clusters = titanic['embarked'].unique()
selected_cluster = random.choice(clusters)

# Take all passengers from the randomly selected cluster
cluster_sample = titanic[titanic['embarked'] == selected_cluster]

print(f"Cluster selected: {selected_cluster}")
print(cluster_sample.head())
```


**Explanation**:

* Randomly selects one group (cluster), such as passengers who boarded from port 'C'.
* All individuals in that cluster are included.


---

## ✅ **4. Systematic Sampling**


```python
# Sort and select every 10th passenger
titanic_sorted = titanic.sort_values(by='age').reset_index(drop=True)
systematic_sample = titanic_sorted.iloc[::10]  # every 10th passenger

print(systematic_sample.head())
```

**Explanation**:

* Starts from the first entry and picks every 10th record.
* Assumes no periodic pattern in ordering.


---

## ✅ **5. Convenience Sampling**


```python
# First 100 rows (those easiest to access)
convenience_sample = titanic.head(100)

print(convenience_sample.head())

```



**Explanation**:

* Fast and easy, but prone to bias.
* Doesn’t guarantee representativeness.


---

## ✅ **6. Judgmental (Purposive) Sampling**


```python
# Select only first-class passengers (assumed important group)
judgmental_sample = titanic[titanic['class'] == 'First'].sample(n=50, random_state=42)

print(judgmental_sample.head())
```

**Explanation**:

* Researcher deliberately selects a group of interest (e.g., wealthier passengers).
* Useful in targeted studies but subjective.

--------
## ✅ **7. Snowball Sampling**

Simulating snowball sampling requires a networked structure, which Titanic doesn’t have. But we can simulate it like:


```python
# Start with a few passengers with same last name (like a family)
initial = titanic[titanic['who'] == 'man'].sample(1)
name_group = titanic[titanic['sibsp'] > 0]  # People with siblings (proxy for network)

# Add related group
snowball_sample = pd.concat([initial, name_group.sample(5, random_state=42)])

print(snowball_sample.head())
```

**Explanation**:

* Starts with one person, then expands through their connections (e.g., family or network).

### 📘 **What is Distribution Theory?**

**Distribution theory** is a fundamental area of statistics and probability that deals with **mathematical descriptions of how data values (random variables) are spread or distributed**. It forms the theoretical backbone for understanding randomness, probability, and statistical inference.

---

## 🔍 **Core Idea of Distribution Theory**

At its core, distribution theory answers these questions:

* **What values can a variable take?**
* **With what probability do those values occur?**

It provides **probability distributions**—formulas or functions that describe the likelihood of all possible outcomes of a random variable (discrete or continuous).

---

## 🧠 **Why Is Distribution Theory Important?**

Distribution theory helps us:

* Describe and model **real-world phenomena**.
* Perform **statistical inference** (estimate population parameters, test hypotheses).
* Develop algorithms in **machine learning** and **bioinformatics**.
* Understand the **behavior of sample data** (via the Central Limit Theorem and Law of Large Numbers).

---

## 🎲 **Types of Distributions in Distribution Theory**

Distribution theory distinguishes between two broad categories:

### 1. **Discrete Distributions**

Used when the variable can take **specific, countable values** (like 0, 1, 2…).

| Distribution  | Description                         | Example in Bioinformatics              |
| ------------- | ----------------------------------- | -------------------------------------- |
| **Bernoulli** | 0 or 1 outcomes                     | Mutation present (1) or absent (0)     |
| **Binomial**  | Number of successes in n trials     | Number of reads with a mutation in DNA |
| **Poisson**   | Count of events in a fixed interval | Number of mutations per gene region    |

---

### 2. **Continuous Distributions**

Used when the variable can take **any value in a range**.

| Distribution          | Description                     | Example in Bioinformatics           |
| --------------------- | ------------------------------- | ----------------------------------- |
| **Normal (Gaussian)** | Bell-shaped; most common        | Gene expression levels              |
| **Exponential**       | Time between events             | Time until next transcription event |
| **Uniform**           | Equal likelihood for all values | Randomly generated quality scores   |

---

## 📐 **Key Concepts in Distribution Theory**

| Concept                                    | Meaning                                                                                      |
| ------------------------------------------ | -------------------------------------------------------------------------------------------- |
| **Probability Density Function (PDF)**     | For continuous variables; shows how likely different outcomes are.                           |
| **Probability Mass Function (PMF)**        | For discrete variables; gives probability for each possible value.                           |
| **Cumulative Distribution Function (CDF)** | Shows the probability that a variable is less than or equal to a value.                      |
| **Moments**                                | Includes mean, variance, skewness—used to describe the shape and spread of the distribution. |

---

## 🧬 **Example in Bioinformatics**

> **Use case**: Suppose we’re studying the number of mutations in a tumor genome.

* If we model it using a **Poisson distribution**, we assume:

  * Mutations happen independently.
  * The average mutation rate is constant per gene.

> If instead we observe that most genes have no mutations but a few have many, we might switch to a **negative binomial** or **zero-inflated** distribution model.

---

## 📊 **Why It Matters for Data Science and Bioinformatics**

* Ensures **appropriate statistical modeling** of complex biological data.
* Guides **simulation studies** and **null distribution** generation.
* Supports **machine learning algorithms** that assume or derive from known distributions (e.g., Naive Bayes assumes Gaussian distribution for features).
