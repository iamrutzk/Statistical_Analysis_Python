# 🧪 **Hypothesis Testing**

---

## 🔍 **Definition**

**Hypothesis Testing** is a **statistical method** used to make decisions about population parameters based on sample data.
It involves evaluating two competing hypotheses — a **null hypothesis ($H_0$)** and an **alternative hypothesis ($H_a$)** — and determining whether the sample data provides enough evidence to **reject the null hypothesis**.

---

## 🧩 **Core Concepts**

### 1. **Null Hypothesis ($H_0$)**

> A default statement that assumes **no effect**, **no difference**, or **status quo**.

📌 *Example:*
$H_0$: "There is no difference in average exam scores between Group A and Group B."

---

### 2. **Alternative Hypothesis ($H_a$)**

> A statement that contradicts the null hypothesis. It suggests that there **is an effect**, **difference**, or **change**.

📌 *Example:*
$H_a$: "Group A has a higher average score than Group B."

---

### 3. **Significance Level ($\alpha$)**

> The threshold probability used to decide whether to reject $H_0$.
> Common values: **0.05**, **0.01**, or **0.10**.

📌 *Interpretation:*
If $\alpha = 0.05$, there's a 5% chance of rejecting $H_0$ when it's actually true.

---

### 4. **P-value**

> The **probability** of obtaining results at least as extreme as the observed data, assuming $H_0$ is true.

* **If p-value ≤ $\alpha$** → reject $H_0$
* **If p-value > $\alpha$** → do not reject $H_0$

---

## 📊 **Step-by-Step Example**

### 🎯 **Scenario:**

A company introduces a **new employee training program** and wants to test if it **improves productivity**.

---

### ✅ **Step 1: Define Hypotheses**

* $H_0$: The training program has **no effect** on productivity.
* $H_a$: The training program **increases** productivity.

---

### ✅ **Step 2: Collect Data**

* Measure productivity scores **before and after** training for 30 employees.

---

### ✅ **Step 3: Choose Significance Level**

* Let $\alpha = 0.05$

---

### ✅ **Step 4: Conduct Statistical Test**

* Use a **paired t-test** to compare before and after scores.

---

### ✅ **Step 5: Analyze Results**

* Suppose the test gives a **p-value = 0.02**

---

### ✅ **Step 6: Decision**

* Since **p-value = 0.02 < α = 0.05**, we **reject $H_0$**.

---

### ✅ **Conclusion**

> The data provides **statistically significant evidence** that the training program increases employee productivity.

---

# 🎯 Steps of Hypothesis Testing

---

## ✅ Step 1: **State the Hypotheses**

You begin by formulating two opposing hypotheses:

* **Null Hypothesis ($H_0$)**: Assumes no effect or no difference.
* **Alternative Hypothesis ($H_a$)**: Assumes an effect or difference exists.

🔸 *Example:*
We want to test if the **average height of men and women** is different.

* $H_0$: $\mu_{men} = \mu_{women}$ (no difference)
* $H_a$: $\mu_{men} \ne \mu_{women}$ (difference exists)

---

## ✅ Step 2: **Collect and Prepare Data**

Collect relevant, representative sample data using appropriate sampling techniques.

🔸 *Example:*
Sample 100 men and 100 women. Record their heights (in cm).

---

## ✅ Step 3: **Choose a Statistical Test**

Choose based on data type, number of groups, and assumptions.

🔹 **Common Tests:**

* **t-test**: Compare two means (used here)
* **Chi-square test**: Categorical association
* **ANOVA**: Compare more than two means

🔸 *Example:*
Use an **independent two-sample t-test** to compare mean heights.

---

## ✅ Step 4: **Calculate Test Statistic and p-value**

Compute the **test statistic** (like a t-score or z-score) and the **p-value** using statistical software or formulas.

* **Test statistic** quantifies difference between groups.
* **p-value** tells how likely that difference is, assuming $H_0$ is true.

🔸 *Example Output:*
t = 2.45, p = 0.015

---

## ✅ Step 5: **Make a Decision**

Compare the p-value to your significance level $\alpha$ (commonly 0.05):

* **If** $p \leq \alpha$: Reject $H_0$
* **If** $p > \alpha$: Do not reject $H_0$

🔸 *Example:*
p = 0.015 < 0.05 → **Reject $H_0$**
→ Evidence suggests a significant difference in height.

---

## ✅ Step 6: **Report and Interpret Results**

Summarize what you found in context:

* Mention the hypotheses.
* State the test used and p-value.
* Interpret what it means practically.

🔸 *Example Summary:*

> “An independent two-sample t-test was conducted to compare heights of men and women. The results showed a significant difference (t = 2.45, p = 0.015). We reject the null hypothesis and conclude that there is a statistically significant difference in average height between men and women.”

# 🧪 Types of Hypothesis Tests

Hypothesis testing methods can be broadly classified into:

* **Parametric Tests** (assume data follows a specific distribution)
* **Non-parametric Tests** (distribution-free tests)

In this section, we’ll focus on **Parametric Tests**, which are widely used when the assumptions are met.

---

## 1️⃣ Parametric Tests

### ✅ Assumptions:

1. **Normality:** Data is normally distributed.
2. **Independence:** Observations are independent.
3. **Homoscedasticity:** Equal variance across groups.
4. **Scale of Measurement:** Data is interval or ratio scaled.

---

## 📌 Common Parametric Tests:

| Test       | Purpose                                                                        |
| ---------- | ------------------------------------------------------------------------------ |
| **t-test** | Compare means of 1 or 2 groups                                                 |
| **ANOVA**  | Compare means across 3 or more groups                                          |
| **Z-test** | Similar to t-test, used when population variance is known and/or large samples |

---

## 1. 🔹 t-Tests

Used when population variance is **unknown**, and sample size is **small (n < 30)**.

---

### a. One-Sample t-Test

* **Used for:** Comparing a sample mean to a known population mean.

**Formula:**

$$
t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}
$$

Where:

* $\bar{x}$ = Sample mean
* $\mu_0$ = Population mean
* $s$ = Sample standard deviation
* $n$ = Sample size

---

### b. Independent Two-Sample t-Test

* **Used for:** Comparing means of two **independent** groups.

**Formula:**

$$
t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{s_p^2 \left( \frac{1}{n_1} + \frac{1}{n_2} \right)}}
$$

Where:

* $s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}$ = Pooled variance
* $n_1, n_2$ = Sample sizes
* $s_1^2, s_2^2$ = Sample variances

---

### c. Paired t-Test

* **Used for:** Comparing two **related** groups (e.g., before vs. after treatment).

**Formula:**

$$
t = \frac{\bar{d}}{s_d / \sqrt{n}}
$$

Where:

* $\bar{d}$ = Mean difference between paired observations
* $s_d$ = Standard deviation of the differences
* $n$ = Number of pairs

---

## 2. 🔹 ANOVA (Analysis of Variance)

Used when comparing means across **3 or more groups**.

### a. One-Way ANOVA

**F-statistic:**

$$
F = \frac{MSB}{MSW}
$$

Where:

* **MSB (Mean Square Between):**

$$
MSB = \frac{\sum_{i=1}^k n_i (\bar{x}_i - \bar{x})^2}{k - 1}
$$

* **MSW (Mean Square Within):**

$$
MSW = \frac{\sum_{i=1}^k \sum_{j=1}^{n_i} (x_{ij} - \bar{x}_i)^2}{N - k}
$$

* $k$: Number of groups
* $n_i$: Size of group $i$
* $\bar{x}_i$: Mean of group $i$
* $\bar{x}$: Overall mean
* $N$: Total observations

---

## 3. 🔹 Z-Tests

Used when:

* Sample size is large ($n > 30$)
* Population variance $\sigma^2$ is known

### a. One-Sample Z-Test

$$
Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}
$$

### b. Two-Sample Z-Test

$$
Z = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}}
$$

---

## 🧠 Example – Independent t-Test

**Scenario:**
A company wants to compare the average monthly sales in two regions: Region A and Region B.

* **$H_0$:** $\mu_1 = \mu_2$ (no difference in sales)
* **$H_a$:** $\mu_1 \ne \mu_2$ (difference exists)

**Data:**

* Region A: $n_1 = 30, \bar{x}_1 = 5200, s_1 = 800$
* Region B: $n_2 = 30, \bar{x}_2 = 4700, s_2 = 850$

### Step 1: Calculate pooled variance:

$$
s_p^2 = \frac{(30 - 1)(800)^2 + (30 - 1)(850)^2}{30 + 30 - 2}
= \frac{29(640000) + 29(722500)}{58}
= \frac{18560000 + 20952500}{58}
= \frac{39512500}{58} \approx 681250
$$

### Step 2: Calculate t-statistic:

$$
t = \frac{5200 - 4700}{\sqrt{681250 \left( \frac{1}{30} + \frac{1}{30} \right)}}
= \frac{500}{\sqrt{681250 \cdot \frac{2}{30}}}
= \frac{500}{\sqrt{45416.67}} \approx \frac{500}{213.1} \approx 2.35
$$

### Step 3: Decision

* Degrees of freedom: $n_1 + n_2 - 2 = 58$
* From t-table, critical value for $\alpha = 0.05$ (two-tailed) ≈ 2.001

✅ Since $2.35 > 2.001$, **reject $H_0$**.
📌 *Conclusion:* There is a statistically significant difference in average sales between Region A and B.

## 🔢 Formula for Chi-Square Test

The general formula for the Chi-Square ($\chi^2$) statistic is:

$$
\chi^2 = \sum \frac{(O - E)^2}{E}
$$

Where:

* $O$ = Observed frequency (what you measured)
* $E$ = Expected frequency (what you expect under the null hypothesis)
* The sum is taken over all categories or cells

---

## 🧪 Steps for Performing a Chi-Square Test of Independence

### Step 1: **State Hypotheses**

* $H_0$: The two categorical variables are independent.
* $H_a$: The two categorical variables are associated (not independent).

---

### Step 2: **Create a Contingency Table**

|           | Category 1  | Category 2  | Total       |
| --------- | ----------- | ----------- | ----------- |
| Group A   | O₁₁         | O₁₂         | Row 1 total |
| Group B   | O₂₁         | O₂₂         | Row 2 total |
| **Total** | Col 1 total | Col 2 total | Grand Total |

---

### Step 3: **Calculate Expected Frequencies**

$$
E_{ij} = \frac{(\text{Row Total}_i \times \text{Column Total}_j)}{\text{Grand Total}}
$$

Do this for each cell.

---

### Step 4: **Calculate Chi-Square Statistic**

$$
\chi^2 = \sum \frac{(O_{ij} - E_{ij})^2}{E_{ij}}
$$

Compute this for all cells and sum them up.

---

### Step 5: **Determine Degrees of Freedom**

$$
df = (r - 1)(c - 1)
$$

Where:

* $r$: Number of rows
* $c$: Number of columns

---

### Step 6: **Compare with Critical Value or Use p-value**

* Use a Chi-square distribution table or calculator.
* If $\chi^2$ > critical value or p-value < 0.05, reject $H_0$.

---

## ✅ Example: Is Voting Preference Related to Gender?

|            | Votes Party X | Votes Party Y | Total |
| ---------- | ------------- | ------------- | ----- |
| **Male**   | 30            | 10            | 40    |
| **Female** | 20            | 20            | 40    |
| **Total**  | 50            | 30            | 80    |

---

### Step 1: Hypotheses

* $H_0$: Gender and voting preference are independent.
* $H_a$: Gender and voting preference are not independent.

---

### Step 2: Calculate Expected Frequencies

$$
E_{\text{Male, X}} = \frac{40 \times 50}{80} = 25  
\quad
E_{\text{Male, Y}} = \frac{40 \times 30}{80} = 15
$$

$$
E_{\text{Female, X}} = \frac{40 \times 50}{80} = 25  
\quad
E_{\text{Female, Y}} = \frac{40 \times 30}{80} = 15
$$

---

### Step 3: Calculate $\chi^2$

$$
\chi^2 = \frac{(30 - 25)^2}{25} + \frac{(10 - 15)^2}{15} + \frac{(20 - 25)^2}{25} + \frac{(20 - 15)^2}{15}
$$

$$
\chi^2 = \frac{25}{25} + \frac{25}{15} + \frac{25}{25} + \frac{25}{15} = 1 + 1.67 + 1 + 1.67 = 5.34
$$

---

### Step 4: Degrees of Freedom

$$
df = (2 - 1)(2 - 1) = 1
$$

---

### Step 5: Decision

* Critical value at $\alpha = 0.05$, df = 1 is **3.84**
* Since $\chi^2 = 5.34 > 3.84$, we **reject $H_0$**

✅ **Conclusion**: There is a significant relationship between gender and voting preference.
