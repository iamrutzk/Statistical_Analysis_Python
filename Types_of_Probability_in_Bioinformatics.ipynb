
# 🧬 Types of Probability in Bioinformatics – Python Examples

This notebook demonstrates key types of probability used in bioinformatics


## 🔹 1. Classical Probability

**📘 Description:** Choosing a random nucleotide from a DNA sequence.

```python
# DNA has 4 nucleotides: A, T, G, C
favorable_outcomes = 1  # e.g., picking an 'A'
total_outcomes = 4
classical_prob = favorable_outcomes / total_outcomes
print(f"Probability of randomly selecting 'A' from DNA: {classical_prob}")
```

**🧠 Insight:** Assumes all nucleotides are equally probable.

## 🔹 2. Empirical Probability

**📘 Description:** Gene BRCA1 was expressed in 120 out of 200 patient samples.

```python
gene_expressed = 120
total_samples = 200
empirical_prob = gene_expressed / total_samples
print(f"Empirical probability that BRCA1 is expressed: {empirical_prob}")
```

**🧠 Insight:** Useful when analyzing lab or sequencing data.

## 🔹 3. Subjective Probability

**📘 Description:** Expert belief about mutation causing disease.

```python
subjective_prob = 0.7
print(f"Subjective probability that the mutation causes disease: {subjective_prob}")
```

**🧠 Insight:** Used when no direct data exists, e.g., hypothesis generation.

## 🔹 4. Conditional Probability

**📘 Description:** Probability a patient has cancer given a BRCA1 mutation.

```python
P_mutation = 0.10
P_cancer_and_mutation = 0.08
P_cancer_given_mutation = P_cancer_and_mutation / P_mutation
print(f"Conditional probability of cancer given BRCA1 mutation: {P_cancer_given_mutation}")
```

**🧠 Insight:** Crucial in genetic risk modeling and clinical genomics.

## 🔹 5. Joint Probability

**📘 Description:** Probability of both BRCA1 and BRCA2 mutations.

```python
P_BRCA1 = 0.10
P_BRCA2 = 0.05
joint_prob = P_BRCA1 * P_BRCA2  # Assuming independence
print(f"Joint probability of both BRCA1 and BRCA2 mutations: {joint_prob}")
```

**🧠 Insight:** Models co-occurrence of mutations.

## 🔹 6. Marginal Probability

**📘 Description:** Mutation probability across all cancer types.

```python
P_mutation_lung = 0.02
P_mutation_breast = 0.03
P_mutation_colon = 0.01
P_total_mutation = P_mutation_lung + P_mutation_breast + P_mutation_colon
print(f"Marginal probability of mutation in any cancer type: {P_total_mutation}")
```

**🧠 Insight:** Used to compute overall mutation rates across categories.

## 🔹 7. Bayesian Probability

**📘 Description:** Disease risk after testing positive.

```python
P_disease = 0.01
P_pos_given_disease = 0.99
P_pos_given_no_disease = 0.05
P_no_disease = 1 - P_disease

P_pos = (P_pos_given_disease * P_disease) + (P_pos_given_no_disease * P_no_disease)
P_disease_given_pos = (P_pos_given_disease * P_disease) / P_pos

print(f"Bayesian probability of disease given positive result: {P_disease_given_pos:.4f}")
```

**🧠 Insight:** Foundation for diagnostic models and risk calculators.

## 🔹 8. Complementary Probability

**📘 Description:** If a gene has 30% chance of being downregulated, what’s the chance it’s not?

```python
P_downregulated = 0.3
P_not_downregulated = 1 - P_downregulated
print(f"Probability gene is not downregulated: {P_not_downregulated}")
```

\**🧠 Insight:** Helps calculate inverse likelihoods in expression studies.

## 🔹 9. Independent Events

**📘 Description:** Two genes are expressed independently.

```python
P_gene1 = 0.4
P_gene2 = 0.6
P_both_expressed = P_gene1 * P_gene2
print(f"Probability both genes are expressed independently: {P_both_expressed}")
```

**🧠 Insight:** Useful in modeling gene co-expression networks.

## 🔹 10. Compound Probability

**📘 Description:** Probability of having mutation A or B.

```python
P_mutA = 0.2
P_mutB = 0.3
P_both = 0.1  # Overlap (intersection)
P_union = P_mutA + P_mutB - P_both
print(f"Probability of having mutation A or B: {P_union}")
```

**🧠 Insight:** Critical in disease panel interpretation and genomic overlap.
