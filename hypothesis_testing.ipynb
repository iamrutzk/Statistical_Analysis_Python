# Introduction to Hypothesis Testing

## Definition and Purpose
Hypothesis testing is a **statistical procedure** used to evaluate assumptions or hypotheses about a population parameter. It is a systematic process that involves:

1. **Formulating hypotheses:**
   - **Null Hypothesis (\(H_0\)):** Represents no effect or no difference.  
     Example: "There is no difference in average test scores between groups A and B."
   - **Alternative Hypothesis (\(H_a\)):** Represents an effect or a difference.  
     Example: "Group A has a higher average test score than group B."

2. **Collecting data:**
   - Data is collected through sampling to represent the population under study.

3. **Evaluating evidence:**
   - Statistical tests are performed to determine whether the evidence from the sample data is strong enough to reject the null hypothesis.

## Purpose of Hypothesis Testing
The primary goal is to make informed inferences about a population based on sample data. This method helps in:
- Quantifying the likelihood that observed differences or patterns are due to random chance.
- Supporting decision-making with statistical rigor.

> **Key Insight:**  
> Hypothesis testing does not "prove" a hypothesis; instead, it evaluates whether the data provides sufficient evidence to reject the null hypothesis.

## Example for Clarification
Suppose a company launches a new training program and wants to test if it significantly improves employee productivity:
- \(H_0\): The training program has no effect on productivity.
- \(H_a\): The training program increases productivity.

Data is collected before and after the training, and a statistical test is used to determine whether the observed increase is statistically significant.


# Steps of Hypothesis Testing

Let’s walk through how to perform a hypothesis test, one step at a time.

---

## Step 1: State Your Hypotheses
The first step is to formulate your research question into two competing hypotheses:

- **Null Hypothesis (\(H_0\)):** The default assumption that there is no effect or difference. It represents the idea that nothing has changed or that there is no significant relationship between variables.  
- **Alternative Hypothesis (\(H_a\)):** The hypothesis that there is an effect or difference. It represents the idea that something has changed or that a relationship exists.

### Example:
Suppose you're testing whether there is a difference in the mean height of men and women.

- **\(H_0\):** The mean height of men is equal to the mean height of women.  
- **\(H_a\):** The mean height of men is not equal to the mean height of women.

Here, the null hypothesis assumes no difference, while the alternative hypothesis suggests a difference exists.

---

## Step 2: Collect and Prepare Data
After stating your hypotheses, you need to collect data to test them. The data should be gathered through experiments, surveys, or observational studies. It is important that the data collection method is designed to test your hypotheses and is representative of the population you want to study.

### Key Considerations:
- **Define the population of interest:** Who or what is the subject of your hypothesis test (e.g., men and women in a population)?  
- **Select an appropriate sampling method:** Random sampling ensures that your sample represents the population well.  
- **Determine the sample size:** A larger sample size generally leads to more reliable results.  
- **Collect and organize the data:** Ensure data is organized systematically, with clear identification of variables.

### Example:
You collect data on the height of 100 men and 100 women from a population to test the hypothesis that their mean heights are equal.

---

## Step 3: Choose the Appropriate Statistical Test
You must choose the right statistical test based on the type of data and the hypothesis. The choice depends on several factors, including:

- **Data type:** Whether the data is continuous (e.g., height, weight) or categorical (e.g., gender, type of product).  
- **Distribution of the data:** Whether the data follows a normal distribution or not.  
- **Sample size:** Whether you have a small or large sample size can influence your choice of test.  
- **Number of groups being compared:** Are you comparing one group, two groups, or multiple groups?

### Common Statistical Tests:
- **t-test (for comparing means):** Used when comparing the means of two groups (e.g., comparing the mean height of men vs. women).  
- **Chi-square test (for categorical data):** Used when examining the relationship between two categorical variables (e.g., testing whether gender is associated with voting preference).  
- **ANOVA (for comparing means of multiple groups):** Used when comparing the means of three or more groups (e.g., testing the average height in multiple regions).  

### Example:
If you're comparing the mean height of men and women, you'd use a **t-test**. If you wanted to compare the mean height across three different regions, you would use **ANOVA**.

---

## Step 4: Calculate the Test Statistic and p-Value
Once you have selected the test, the next step is to calculate the **test statistic** and the **p-value**.

- The **test statistic** quantifies how much the sample data deviates from the null hypothesis. This is typically calculated using a formula or statistical software.
- The **p-value** represents the probability of observing results as extreme as the sample data, assuming the null hypothesis is true.

The p-value helps you determine whether the observed data provides sufficient evidence to reject the null hypothesis.

### Example:
For a **t-test**, you would calculate a t-statistic based on your sample data (e.g., sample means, standard deviation) and compare it to a critical value from the t-distribution. The p-value is then computed based on this t-statistic.

---

## Step 5: Make a Decision
After calculating the p-value, compare it to a predetermined significance level (\( \alpha \)), which is often set at **0.05**. This significance level defines the threshold for rejecting the null hypothesis.

### Decision Rule:
- **If p-value ≤ α (e.g., 0.05):** Reject the null hypothesis. This suggests that the evidence supports the alternative hypothesis.
- **If p-value > α:** Fail to reject the null hypothesis. This suggests that there is insufficient evidence to support the alternative hypothesis.

> **Important Note:** Failing to reject the null hypothesis does not "prove" it is true. It simply means there is not enough evidence to conclude that the null hypothesis is false.

### Example:
If you calculate a p-value of **0.03**, which is less than **0.05**, you would reject the null hypothesis and conclude that the mean height of men is significantly different from that of women.

---

## Step 6: Present Your Findings
Finally, once you have made a decision, you need to present your findings in a clear and concise manner. This typically involves:

1. **Reporting the results:** Include the test statistic, p-value, and your decision regarding the null hypothesis.
2. **Contextualizing the findings:** Explain what the results mean in the context of your study.  
3. **Practical significance:** Discuss whether the observed effect is practically important, not just statistically significant.

### Key Points to Include:
- **Context for the study:** What was being tested and why?  
- **State the hypotheses:** Clearly state both \(H_0\) and \(H_a\).  
- **Report the test statistic and p-value:** E.g., "The t-test statistic was 2.5 with a p-value of 0.03."  
- **Interpret the results:** "Since the p-value is less than 0.05, we reject the null hypothesis, concluding that the mean height of men is significantly different from the mean height of women."

---

By following these steps, you can systematically evaluate hypotheses and draw meaningful conclusions from your data.



# Types of Hypothesis Tests

Hypothesis tests can be broadly categorized into two main types: **Parametric tests** and **Non-parametric tests**. These tests are chosen based on the characteristics of the data and the assumptions they meet.

---

## 1. Parametric Tests
Parametric tests assume that the data follows a specific probability distribution, typically the **normal distribution**. These tests are generally more powerful when the assumptions are met, meaning they are more likely to detect a true effect or difference.

### Key Assumptions of Parametric Tests:
1. The data is normally distributed.  
2. The sample data is collected independently.  
3. The variances in the groups being compared are equal (homoscedasticity).  
4. The data is measured on an interval or ratio scale.

---

### Common Parametric Tests:
1. **t-tests**
2. **ANOVA**
3. **Z-tests**

---

### 2. ANOVA (Analysis of Variance)
ANOVA is used to compare means across multiple groups.

#### **a. One-way ANOVA**
- **Purpose:** Tests whether the means of three or more groups are equal.  
- **Formula for F-statistic:**

\[
F = \frac{\text{Between-group variability (MSB)}}{\text{Within-group variability (MSW)}}
\]

Where:
- **MSB (Mean Square Between):**

\[
MSB = \frac{\sum_{i=1}^k n_i (\bar{x}_i - \bar{x})^2}{k - 1}
\]

- **MSW (Mean Square Within):**

\[
MSW = \frac{\sum_{i=1}^k \sum_{j=1}^{n_i} (x_{ij} - \bar{x}_i)^2}{N - k}
\]

Here:
- \( k \): Number of groups  
- \( n_i \): Sample size of group \( i \)  
- \( \bar{x}_i \): Mean of group \( i \)  
- \( \bar{x} \): Overall mean  
- \( N \): Total sample size  

---

### **1. t-Test Use Case**
**Scenario**:  
A biologist is studying the difference in sepal lengths between two species of flowers (e.g., Setosa and Versicolor) using the Iris dataset.

**Hypothesis**:
- Null (H₀): There is no difference in the mean sepal lengths of Setosa and Versicolor.
- Alternative (H₁): There is a difference in the mean sepal lengths.

**Implementation**:

```python

from scipy.stats import ttest_ind
from sklearn.datasets import load_iris
import pandas as pd

# Load Iris dataset
iris = load_iris(as_frame=True)
df = iris.data
df['species'] = iris.target
# Extract two groups (Setosa and Versicolor)
group1 = df[df['species'] == 0]['sepal length (cm)']
group2 = df[df['species'] == 1]['sepal length (cm)']

# Perform t-test
t_stat, p_value = ttest_ind(group1, group2)

print("t-test results:")
print(f"t-statistic: {t_stat:.3f}, p-value: {p_value:.3f}")

# Conclusion
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```

### **2. z-Test Use Case**
**Scenario**:  
A restaurant owner is analyzing if there's a difference in the average total bill between smokers and non-smokers using the Tips dataset.

**Hypothesis**:
- Null (H₀): The average total bill for smokers equals that of non-smokers.
- Alternative (H₁): The average total bill for smokers is different from non-smokers.

**Implementation**:

---




```python
#pip install statsmodels
```

```python
import seaborn as sns
from statsmodels.stats.weightstats import ztest

# Load Tips dataset
tips = sns.load_dataset('tips')

# Separate groups
smokers = tips[tips['smoker'] == 'Yes']['total_bill']
non_smokers = tips[tips['smoker'] == 'No']['total_bill']

# Perform z-test
z_stat, p_value = ztest(smokers, non_smokers)

print("z-test results:")
print(f"z-statistic: {z_stat:.3f}, p-value: {p_value:.3f}")

# Conclusion
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```


### **3. ANOVA Use Case**
**Scenario**:  
A researcher wants to know if the alcohol content varies among three different classes of wine using the Wine dataset.

**Hypothesis**:
- Null (H₀): The mean alcohol content is the same across all wine classes.
- Alternative (H₁): At least one class has a different mean alcohol content.

**Implementation**:

```python
from scipy.stats import f_oneway
from sklearn.datasets import load_wine

# Load Wine dataset
wine = load_wine(as_frame=True)
df = wine.data
df['target'] = wine.target

# Extract alcohol content for each class
group1 = df[df['target'] == 0]['alcohol']
group2 = df[df['target'] == 1]['alcohol']
group3 = df[df['target'] == 2]['alcohol']

# Perform ANOVA
f_stat, p_value = f_oneway(group1, group2, group3)

print("ANOVA results:")
print(f"F-statistic: {f_stat:.3f}, p-value: {p_value:.3f}")

# Conclusion
if p_value < 0.05:
    print("Reject the null hypothesis: At least one group differs.")
else:
    print("Fail to reject the null hypothesis: No significant difference.")
```

### **4. Chi-Square Test Use Case**
**Scenario**:  
An analyst is investigating the relationship between passenger class and survival status on the Titanic.

**Hypothesis**:
- Null (H₀): Survival is independent of passenger class.
- Alternative (H₁): Survival is associated with passenger class.

**Implementation**:

```python


import seaborn as sns
from scipy.stats import chi2_contingency
import pandas as pd

# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Create contingency table
contingency_table = pd.crosstab(titanic['class'], titanic['survived'])
print(contingency_table)
# Perform Chi-Square test
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square test results:")
print(f"Chi2 statistic: {chi2:.3f}, p-value: {p_value:.3f}, degrees of freedom: {dof}")

# Conclusion
if p_value < 0.05:
    print("Reject the null hypothesis: There is an association between variables.")
else:
    print("Fail to reject the null hypothesis: No association between variables.")
```
