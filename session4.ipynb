
### 🎓 **Introduction to Probability**

**Probability** is a branch of mathematics that deals with measuring the **likelihood** of events happening under conditions of uncertainty. It plays a central role in **data science**, **bioinformatics**, **AI**, **genetics**, **clinical trials**, and many other domains. Understanding probability allows scientists and researchers to make informed predictions, assess risk, and make data-driven decisions.

---

#### 🧠 Why is Probability Important?

* **In Bioinformatics**: It helps assess the likelihood that a gene is differentially expressed, or that a DNA sequence contains a mutation.
* **In Data Science**: Probability underpins algorithms like Naive Bayes, logistic regression, and probabilistic graphical models.
* **In Decision Making**: Whether estimating disease risk or predicting model accuracy, probability is fundamental to drawing conclusions from data.

---

#### 📌 **Key Concepts**

---

### 1. **Sample Space ($\Omega$)**

* The **sample space** is the set of all possible outcomes of a random experiment.
* Denoted by the Greek letter $\Omega$ (omega).

📌 **Example 1: Tossing a Coin**

* Experiment: Tossing a fair coin once
* Sample Space: $\Omega = \{ \text{Heads}, \text{Tails} \}$

📌 **Example 2: Rolling a Die**

* Experiment: Rolling a 6-sided die
* Sample Space: $\Omega = \{ 1, 2, 3, 4, 5, 6 \}$

📌 **Bioinformatics Example:**

* Experiment: Picking a nucleotide from DNA
* Sample Space: $\Omega = \{ A, T, G, C \}$

---

### 2. **Event ($E$)**

* An **event** is any subset of the sample space.
* It can include one or more outcomes.

📌 **Example 1: Rolling an Even Number**

* Sample Space: $\Omega = \{ 1, 2, 3, 4, 5, 6 \}$
* Event $E$: Rolling an even number
  → $E = \{ 2, 4, 6 \}$

📌 **Example 2: Getting a Purine Nucleotide**

* Sample Space: $\Omega = \{ A, T, G, C \}$
* Event $E$: Getting a purine (A or G)
  → $E = \{ A, G \}$

---

### 3. **Probability of an Event ($P(E)$)**

* The **probability** of an event is a number between **0** and **1**:

  * **0** means the event is **impossible**
  * **1** means the event is **certain**
* If all outcomes are **equally likely**, we use the classical probability formula:

$$
P(E) = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}}
$$

📌 **Example 1: Rolling a 4 on a Fair Die**

* Favorable Outcomes: {4}
* Total Outcomes: 6
* Probability: $P(4) = \frac{1}{6} \approx 0.1667$

📌 **Example 2: Selecting a G or C from DNA**

* Sample Space: $\Omega = \{ A, T, G, C \}$
* Event: $E = \{ G, C \}$
* Probability: $P(E) = \frac{2}{4} = 0.5$

---

### 🎯 Applications of Probability

#### 1. **Decision Making**

Helps in choosing actions with the least risk or highest gain based on the likelihood of outcomes.

*Example:* Deciding whether to treat a patient based on the probability of a disease derived from test results.

#### 2. **Gaming & Simulation**

Used in modeling random events like dice rolls, card draws, or genetic recombination.

*Example:* Simulating genotypes using Mendelian inheritance probabilities.

#### 3. **Artificial Intelligence & Machine Learning**

Probability models the uncertainty in predictions. It is key in:

* Naive Bayes Classifier
* Probabilistic Neural Networks
* Markov Chains

*Example:* Predicting if an email is spam based on the probability of certain words appearing.

## 📘 **1.1 What is Probability?**

---

### ✅ **Definition**

**Probability** is the branch of mathematics that deals with the **measurement of uncertainty**. It provides a way to **quantify how likely** an event is to occur, especially when outcomes are not guaranteed.

$$
P(E) = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}} \quad \text{(if all outcomes are equally likely)}
$$

---

### 🧠 **Why Probability?**

In real-world biological and data science problems, uncertainty is everywhere:

* Will a gene be expressed in a certain condition?
* Will a patient develop a disease given a specific mutation?
* Will a classification model predict correctly?

Probability gives us the language and tools to answer such questions **mathematically**.

---

### 🔑 **Key Characteristics of Probability**

1. **Range is Between 0 and 1**:

   * $P(E) = 0$: The event **cannot happen** (impossible).
   * $0 < P(E) < 1$: The event is **uncertain**, but possible.
   * $P(E) = 1$: The event **will definitely happen** (certain).

2. **Sum Rule** (Law of Total Probability):

   * The total probability of **all** mutually exclusive outcomes in the sample space must equal 1.

     $$
     \sum_{i} P(E_i) = 1
     $$

---

### 🔍 **Examples**

#### 🎲 **Example 1: Coin Toss**

* **Experiment**: Tossing a fair coin
* **Sample Space** ($\Omega$): { Heads, Tails }
* **Event $E$**: Getting Heads
* **Favorable outcomes**: 1 (Heads)
* **Total outcomes**: 2 (Heads, Tails)

$$
P(\text{Heads}) = \frac{1}{2} = 0.5
$$

✅ *The chance of getting heads is 50%.*

---

#### 🎲 **Example 2: Rolling a Die**

* **Experiment**: Rolling a standard 6-sided die
* **Sample Space** ($\Omega$): { 1, 2, 3, 4, 5, 6 }
* **Event $E$**: Rolling an even number → $E = \{2, 4, 6\}$
* **Favorable outcomes**: 3
* **Total outcomes**: 6

$$
P(\text{Even number}) = \frac{3}{6} = 0.5
$$

✅ *There is a 50% chance of getting an even number.*

---

#### 🧬 **Example 3: Bioinformatics - DNA Base Selection**

* **Experiment**: Randomly selecting one nucleotide from DNA
* **Sample Space**: $\Omega = \{ A, T, G, C \}$
* **Event $E$**: Selecting a purine (A or G)
* **Favorable outcomes**: 2 (A, G)
* **Total outcomes**: 4

$$
P(\text{Purine}) = \frac{2}{4} = 0.5
$$

✅ *The probability of picking a purine is 0.5.*

---

### 📌 **Importance of Probability**

#### 📈 **In Statistics**:

* Helps design **experiments** and analyze **data under uncertainty**.
* Foundation for **confidence intervals**, **hypothesis testing**, and **Bayesian analysis**.

#### 🧬 **In Bioinformatics**:

* Used in models like **Hidden Markov Models** (e.g., for gene prediction).
* Assesses **probability of mutations**, gene-disease associations.

#### 🤖 **In Data Science & AI**:

* Powers **probabilistic models** like Naive Bayes.
* Used in **prediction**, **classification**, **risk modeling**.

## 🎓 **1.2 Types of Probability**

Understanding different types of probability helps in choosing the right approach for analyzing data, drawing conclusions, and modeling real-world problems in statistics, data science, and bioinformatics.

---

### 1. 🧮 **Classical (Theoretical) Probability**

**Definition:**
Used when all outcomes are equally likely.

$$
P(E) = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}}
$$

**Example:**

* Rolling a 4 on a fair 6-sided die:
  $P(4) = \frac{1}{6}$

**Bioinformatics Context:**

* Picking a random nucleotide from DNA:
  $P(A) = \frac{1}{4}$ (assuming equal distribution of A, T, G, C)

---

### 2. 🧪 **Empirical (Experimental) Probability**

**Definition:**
Based on observed data from experiments or studies.

$$
P(E) = \frac{\text{Number of times event occurred}}{\text{Total number of trials}}
$$

**Example:**

* Tossing a coin 100 times, heads appears 45 times:
  $P(\text{Heads}) = \frac{45}{100} = 0.45$

**Bioinformatics Context:**

* Out of 1,000 patients tested, 120 had a mutation in gene BRCA1:
  $P(\text{BRCA1 mutation}) = \frac{120}{1000} = 0.12$

---

### 3. 🧠 **Subjective Probability**

**Definition:**
Based on personal belief, intuition, or expert judgment — **not data**.

**Example:**

* A doctor estimates there's a 70% chance a patient has a viral infection, based on symptoms and experience.

**Bioinformatics Context:**

* A researcher might believe there's a high chance a gene is involved in cancer, based on early lab results.

---

### 4. 🔗 **Conditional Probability**

**Definition:**
Probability of an event **A**, given that **B has already occurred**.

$$
P(A|B) = \frac{P(A \cap B)}{P(B)}
$$

**Example:**

* Drawing a King given it’s known the card is a face card:
  $P(\text{King} | \text{Face}) = \frac{4}{12}$

**Bioinformatics Context:**

* $P(\text{Disease} | \text{Mutation})$:
  Probability of disease given a known genetic mutation.

---

### 5. 🔄 **Joint Probability**

**Definition:**
Probability of **two or more events happening at the same time**.

$$
P(A \cap B)
$$

**Example:**

* Rolling a 4 and a 6 on two dice:
  $P(4 \text{ and } 6) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36}$

**Bioinformatics Context:**

* Probability of finding **mutation A and mutation B** in the same patient.

---

### 6. 🧩 **Marginal Probability**

**Definition:**
Probability of a **single event**, regardless of other variables.

$$
P(A) = \sum P(A \cap B_i)
$$

**Example:**

* Rolling an even number:
  $P(\text{Even}) = P(2) + P(4) + P(6) = \frac{3}{6} = 0.5$

**Bioinformatics Context:**

* Marginal probability that a sample has a specific mutation, regardless of phenotype or condition.

---

### 7. 🧬 **Bayesian Probability**

**Definition:**
Updates prior beliefs using new evidence — **Bayes' Theorem**.

$$
P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}
$$

**Example:**

* Disease testing:

  * $P(\text{Disease}) = 0.01$
  * Test is 99% sensitive; 5% false positive rate.
  * Given a positive result, what’s the actual chance the patient has the disease?

**Bioinformatics Context:**

* Estimating the probability of a disease **given** a positive test result for a genetic marker.

---

### 8. ♻️ **Complementary Probability**

**Definition:**
Probability that an event **does not happen**.

$$
P(A^c) = 1 - P(A)
$$

**Example:**

* If $P(\text{Rain}) = 0.3$, then
  $P(\text{No Rain}) = 1 - 0.3 = 0.7$

**Bioinformatics Context:**

* If the probability of a gene being upregulated is 0.2, then the probability it's not upregulated is 0.8.

---

### 9. 🔗 **Independent Events**

**Definition:**
Two events are independent if the occurrence of one **does not affect** the other.

$$
P(A \cap B) = P(A) \cdot P(B)
$$

**Example:**

* Rolling two dice:
  $P(4 \text{ and } 5) = \frac{1}{6} \cdot \frac{1}{6} = \frac{1}{36}$

**Bioinformatics Context:**

* Expression of **Gene A** is independent of **Gene B**, so:

  $$
  P(\text{A and B}) = P(A) \cdot P(B)
  $$

---

### 10. 🧮 **Compound Probability**

**Definition:**
Probability involving **combinations** of events — typically unions or intersections.

* **Union (A or B)**:

  $$
  P(A \cup B) = P(A) + P(B) - P(A \cap B)
  $$

* **Intersection (A and B)**:

  $$
  P(A \cap B) = P(A) \cdot P(B|A) \quad \text{(if dependent)}
  $$

**Example:**

* Rolling an even number or a number > 3:

  * $P(\text{Even}) = 3/6$
  * $P(>3) = 3/6$
  * $P(\text{Even} \cap >3) = 2/6$
  * Result:

    $$
    P(\text{Even or >3}) = \frac{3}{6} + \frac{3}{6} - \frac{2}{6} = \frac{4}{6}
    $$

**Bioinformatics Context:**

* Finding the probability that a patient has either **mutation A or B** (or both).
